# Input size and work performed

<div class='big center'>The input size describes the input—not necessarily how much of it the algorithm processes.</div>

<div style='display:flex; gap:24px; align-items:stretch; margin-top:24px;'>

<div style='flex:1; padding:20px; border:2px solid #517fa4; border-radius:12px; background:#f3f7fa;'>
<h3 style='color:#355f7c; margin-top:0;'>An integer input</h3>

For an integer $x$, the input size is the number of bits in its binary representation:

$$n = \lfloor \log_2 x \rfloor + 1.$$

<b>Algorithm 1: Count the 1-bits</b>

```text
countOnes(x):
    count = 0
    while x > 0:
        count += x mod 2
        x = x div 2
    return count
```

This algorithm processes all $n$ bits. The amount of work therefore grows with the input size.

<b>Algorithm 2: Check whether the integer is even</b>

```text
isEven(x):
    return x mod 2 == 0
```

The input still has size $n$, but this algorithm only needs to inspect the last bit. The amount of work remains constant as the input grows.
</div>

<div style='flex:1; padding:20px; border:2px solid #136f63; border-radius:12px; background:#f2faf8;'>
<h3 style='color:#136f63; margin-top:0;'>A matrix input</h3>

```text
sumAll(matrix):
    total = 0
    for each row:
        for each element in the row:
            total += element
    return total
```

For a matrix with $r$ rows and $c$ columns, the algorithm processes $rc$ elements.

For a square matrix, we can describe this work in two ways:

- If $n$ is the number of rows and columns, the algorithm processes $n^2$ elements.
- If $N$ is the total number of elements, the algorithm processes $N$ elements.

These statements describe the same amount of work because $N=n^2$. Using $N$ does not make the algorithm faster—it simply uses a different parameter to describe the input.
</div>

</div>

<div style='margin-top:24px; padding:14px 20px; border-left:6px solid #e0922f; background:#fff8eb;'>
<b>Key rule:</b> Clearly define what each input-size parameter represents before describing how the work grows.

# Step 2 — Identify a primitive operation

A **primitive (basic) operation** is a small operation whose cost we treat as constant in our model.

<table>
<thead><tr><th>Possible primitive operation</th><th>Context</th></tr></thead>
<tbody>
<tr><td><code>a[i] == target</code></td><td>search</td></tr>
<tr><td><code>a[i] &lt; a[j]</code></td><td>sorting</td></tr>
<tr><td><code>sum += a[i]</code></td><td>accumulation</td></tr>
<tr><td>follow one reference</td><td>linked traversal</td></tr>
</tbody>
</table>

<div class='card'>💡 <b>Key idea:</b> Choose the operation that represents the algorithm's repeated work — everything else is either done once, or done alongside that operation a fixed number of times.</div>

# Step 3 — Count it

```java
static int largest(int[] a) {
    int best = 0;
    for (int i = 1; i < a.length; i++)
        if (a[best] < a[i]) best = i;
    return best;
}
```

- Input size: $n=$ `a.length`
- Exact Count: $const1 * (n-1) + const2$, every time — this algorithm has no best/worst-case split

<div class='card'>💡 <b>Key idea:</b> The operation count gives us a cost function $T(n)$.</div>

# Not every count is a single formula

`largest` counts the same way on every input of size $n$. Most algorithms don't:

```java
static boolean contains(int[] a, int target) {
    for (int value : a)
        if (value == target) return true;
    return false;
}
```

Here the comparison count **depends on where `target` is**, not just on $n$. So we'll classify inputs of size $n$ into cases:

<table>
<thead><tr><th>Case</th><th>Question it answers</th></tr></thead>
<tbody>
<tr><td>Worst</td><td>What's the most work on any input of size $n$? (a guarantee)</td></tr>
<tr><td>Best</td><td>What's the least work on any input of size $n$?</td></tr>
<tr><td>Average</td><td>What's the expected work, given a distribution over inputs?</td></tr>
</tbody>
</table>

<div class='warn'>⚠️ <b>Important:</b> Each case is <i>still just a cost function</i> $T(n)$ — worst-case is the most common because it's a guarantee that needs no assumptions about input distribution.</div>

# Step 4 — Keep the dominant growth

<div class='hero center'>
<p>$T(n)=7n^2+40n+900$</p>
<h1>&rArr; Quadratic growth: $n^2$</h1>
</div>

As $n$ becomes sufficiently large:

- $n^2$ grows faster than $n$ and the constant term;
- the highest-order term therefore determines the overall shape of the growth;
- we omit its constant factor and the lower-order terms when describing the growth class.

<div style='margin-top:20px; padding:14px 20px; border-left:Rect6px solid #e0922f; background:#hráfff8eb;'>
<b>Important:</b> The constants and lower-order terms can still affect actual running time, especially for small inputs. We omit them only when studying how the work scales for large inputs.
</div>


# Why can we drop constants?

For $n\ge1$:

$$7n^2 \le 7n^2+40n+900 \le 947n^2$$

The function is trapped between two constant multiples of $n^2$ — that's the entire idea the definitions on the next few slides make precise.

<div class='big center'>Its scale changes; its shape does not.</div>

# When the bound is tight: $\Theta$

If we can trap $T(n)$ between two constant multiples of $g(n)$, we have a **tight asymptotic bound**.

$$T(n)\in\Theta(g(n))$$

Read: "$T(n)$ is theta of $g(n)$."

Interpretation: after some $n_0$, $T(n)$ stays between positive constant multiples of $g(n)$:

$$0\le c_1g(n)\le T(n)\le c_2g(n)\quad(n\ge n_0)$$

# $\Theta$ example 1 — Exact count

```java
for (int i = 0; i < n; i++)
    for (int j = i; j < n; j++)
        compare();
```

$$T(n)=n+(n-1)+\cdots+1=\frac{n^2+n}{2}$$

Therefore:

$$T(n)\in\Theta(n^2+n)=\Theta(n^2)$$

# $\Theta$ example 2 — No exact formula needed

Suppose an algorithm performs $T(n)$ operations and we can prove, for $n\ge n_0$:

$$2(n\log n+n)\le T(n)\le9(n\log n+n)$$

We may never pin down the exact formula, but both sides have the same shape.

<div class='card'>💡 <b>Key idea:</b> $T(n)\in\Theta(n\log n+n)=\Theta(n\log n)$ — Θ only requires trapping $T(n)$, not counting it exactly.</div>

# When we only have an upper bound: $O$

$$T(n)\in O(g(n))$$

Read: "$T(n)$ is big-O of $g(n)$."

Interpretation: beyond some $n_0$, $T(n)$ grows **no faster than** a constant multiple of $g(n)$:

$$0\le T(n)\le c\,g(n)\quad(n\ge n_0)$$

<div class='warn'>⚠️ <b>Important:</b> $O$ does not mean "exactly," and it does not automatically mean "worst case" — you can take the $O$ of a best-case, worst-case, or average-case function.</div>

# Big-O example — A guaranteed ceiling

A routine performs no more than

$$4n^2+7n+3$$

operations. Choose $g(n)=n^2+n+1$. For $n\ge1$:

$$4n^2+7n+3\;\le\;7(n^2+n+1)$$

<div class='card'>💡 <b>Key idea:</b> $T(n)\in O(n^2+n+1)$ — this guarantees a ceiling; it says nothing about whether that ceiling is ever actually reached.</div>

# When we only have a lower bound: $\Omega$

$$T(n)\in\Omega(g(n))$$

Read: "$T(n)$ is big-Omega of $g(n)$."

Interpretation: beyond some $n_0$, $T(n)$ grows **at least as fast as** a constant multiple of $g(n)$:

$$0\le c\,g(n)\le T(n)\quad(n\ge n_0)$$

<div class='warn'>⚠️ <b>Important:</b> $\Omega$ does not automatically mean "best case" — the same caution as $O$ applies in reverse.</div>

# Big-Omega example — Unavoidable work

To find the maximum in an unsorted array, every correct algorithm must inspect all $n$ values in the worst case — skip one, and that one might have been the maximum.

Suppose inspection plus loop control needs at least $2n+1$ primitive operations. With $g(n)=n+1$, for $n\ge1$:

$$T(n)\ge2n+1\ge g(n)$$

<div class='card'>💡 <b>Key idea:</b> $T(n)\in\Omega(n+1)=\Omega(n)$ — a lower bound is a claim about every possible algorithm for the problem, not just the one we wrote.</div>

# Standard growth classes

The notations $O$, $\Omega$, and $\Theta$ all use the same common growth functions:

$$1\prec\log n\prec n\prec n\log n\prec n^2\prec 2^n$$

<div style='text-align:center; color:#555;'>Each function can describe an upper bound ($O$), a lower bound ($\Omega$), or a tight bound ($\Theta$).</div>


# A bound need not be tight

If $T(n)=3n+2$, all four statements below are **true**:

- $T(n)\in O(n)$ — tight as an upper bound
- $T(n)\in O(n^2)$ — true, but loose
- $T(n)\in\Omega(1)$ — true, but loose
- $T(n)\in\Theta(n)$ — the most informative statement we can make

<div class='card'>💡 <b>Key idea:</b> An algorithm has infinitely many valid upper and lower bounds. Convention: state the tightest one you can justify — that's why $\Theta$ is preferred whenever it's known.</div>

# Recognize common loop patterns

<div class='compare'><div><pre>for (int i=0; i&lt;n; i++)
    work();</pre><b>$\Theta(n)$</b></div><div><pre>for (int i=1; i&lt;n; i*=2)
    work();</pre><b>$\Theta(\log n)$</b></div></div>

- Sequential blocks: **add** their costs.
- Independent nested loops (inner bound doesn't depend on outer variable): **multiply** iteration counts.
- Dependent nested loops (inner bound depends on outer variable): **write a sum**.

# Dependent nested loops

```java
for (int i = 0; i < n; i++)
    for (int j = 0; j <= i; j++)
        work();
```

$$1+2+\cdots+n=\frac{n(n+1)}2\in\Theta(n^2)$$

<div class='warn'>⚠️ <b>Important:</b> Do not multiply loop bounds blindly here — the inner loop runs $i+1$ times, not $n$ times. Multiplying would wrongly suggest $\Theta(n^2)$ is obvious rather than derived.</div>

# Recursive complexity: write a recurrence

```java
static int sum(int[] a, int n) {
    if (n == 0) return 0;
    return a[n - 1] + sum(a, n - 1);
}
```

$$T(n)=T(n-1)+\Theta(1)$$

Expand: $T(n)=T(n-1)+c=T(n-2)+2c=\cdots=T(0)+nc$.

<div class='card'>💡 <b>Key idea:</b> $n$ levels of recursion × constant work per level = $\Theta(n)$ time.</div>

<div class='card'>💡 <b>Key idea:</b> Use the simplest method that makes the repeated work visible — reach for the Master Theorem once the recurrence has divide-and-conquer's shape.</div>

# Divide and combine

$$T(n)=2T(n/2)+\Theta(n)$$

Recursion tree: $\log_2 n$ levels, and total work per level is $\Theta(n)$ &nbsp;&rarr;&nbsp; $\Theta(n\log n)$.

**Master Theorem**, for $T(n)=aT(n/b)+f(n)$, compare $f(n)$ to $n^{\log_b a}$:

<table>
<thead><tr><th>Compare $f(n)$ to $n^{\log_b a}$</th><th>Result</th></tr></thead>
<tbody>
<tr><td>$f(n)$ grows slower</td><td>$\Theta(n^{\log_b a})$ — the leaves dominate</td></tr>
<tr><td>$f(n)$ grows at the same rate</td><td>$\Theta(n^{\log_b a}\log n)$ — every level ties</td></tr>
<tr><td>$f(n)$ grows faster</td><td>$\Theta(f(n))$ — the top level dominates</td></tr>
</tbody>
</table>

Here $a=2,b=2$, so $n^{\log_b a}=n$, matching $f(n)=\Theta(n)$: the middle case, giving $\Theta(n\log n)$.

# Branching recursion can explode

```java
static int fib(int n) {
    if (n <= 1) return n;
    return fib(n - 1) + fib(n - 2);
}
```

- Recurrence: $T(n)=T(n-1)+T(n-2)+\Theta(1)$ — two subproblems, not one.
- The same smaller subproblems get recomputed over and over, so the number of calls grows exponentially.
- Tight time: $\Theta(\varphi^n)$ where $\varphi\approx1.618$ (the golden ratio); a simpler, looser upper bound is $O(2^n)$.
- Maximum call-stack depth is still only $\Theta(n)$ — depth follows the *longest path* down, not the total number of calls.

# Amortized analysis

Some individual operations are expensive, but they happen rarely. **Amortized analysis** spreads that cost over a whole sequence of operations.

Dynamic-array append with capacity doubling:

# Why dynamic-array copying stays linear

When capacity doubles each time it's exceeded, the copied-element counts across all resizes form a geometric sum:

$$1+2+4+\cdots+\frac n2<n$$

Add the $n$ writes from the appends themselves: total work across the whole sequence is still $\Theta(n)$, so each append is $\Theta(1)$ amortized.

<div class='warn'>⚠️ <b>Important:</b> Amortized is <i>not</i> average case. It needs no probability distribution over inputs — it's a guarantee about the total cost of any sequence of $n$ operations.</div>

# Space complexity

Space analysis asks how memory grows with input size — same tools ($\Theta$, $O$, $\Omega$), applied to memory instead of operation counts.

<table>
<thead><tr><th>Space</th><th>Meaning</th></tr></thead>
<tbody>
<tr><td>Input space</td><td>memory occupied by the input itself</td></tr>
<tr><td>Auxiliary space</td><td>extra memory the algorithm uses beyond the input</td></tr>
</tbody>
</table>

We almost always report **auxiliary** space, since input space is fixed by the problem, not the algorithm.

- A few variables: $\Theta(1)$ auxiliary space.
- A copy of an array of size $n$: $\Theta(n)$.
- Recursion: don't forget the call stack — it's memory too.

# Time and space can disagree

```java
static int sum(int[] a, int n) {
    if (n == 0) return 0;
    return a[n - 1] + sum(a, n - 1);
}
```

- Time: $\Theta(n)$
- Auxiliary space: $\Theta(n)$ — because up to $n$ stack frames coexist at once

An **iterative** sum has the same $\Theta(n)$ time, but only $\Theta(1)$ auxiliary space.

<div class='card'>💡 <b>Key idea:</b> A faster or simpler-looking algorithm isn't automatically cheaper overall — good design weighs time, space, and clarity against what the workload actually needs, not time complexity alone.</div>

# Final checklist

1. Define the input size.
2. Select and count a primitive operation.
3. Decide whether the count depends on more than $n$ — if so, specify the case (best / worst / average).
4. Prove an upper bound, a lower bound, or both; report $\Theta$ when you can justify it, $O$ or $\Omega$ otherwise.
5. Name the standard growth class — that's what makes algorithms comparable.
6. Analyze auxiliary space separately; check whether recursion is involved.
7. If operations vary sharply in cost across a sequence, ask whether an amortized bound is the more honest one.
8. Validate the important constants empirically — Θ/O/Ω predict scalability, not wall-clock time.

# Activity B — Count equal pairs

```java
static int equalPairs(int[] a) {
    int count = 0;
    for (int i = 0; i < a.length; i++)
        for (int j = i + 1; j < a.length; j++)
            if (a[i] == a[j]) count++;
    return count;
}
```

- Give the exact number of equality comparisons.
- Give a tight time bound.
- Does the input's content change the time case, or just the return value?
- Give the auxiliary space complexity.

# Activity C — Recursive processing

```java
static int process(int[] a, int n) {
    if (n <= 1) return a[0];
    int total = 0;
    for (int i = 0; i < n; i++) total += a[i];
    return total + process(a, n / 2);
}
```

- Write the time recurrence and solve it.
- Give an upper, lower, and tight time bound.
- Determine the maximum call-stack space.
