### 3. Install Java and verify the terminal workflow

You will need a **Java Development Kit (JDK)**, not only a Java runtime. The JDK provides the tools used to compile and run Java programs.

- [ ] Install a recent JDK version (JDK 25 recommended): **[link](https://adoptium.net/temurin/releases/?version=25)**
- [ ] Open a terminal.
- [ ] Run `java --version`.
- [ ] Run `javac --version`.
- [ ] Confirm that both commands print version information without an error.

If either command is not recognized, Java may not be installed correctly or your system may not know where to find it. Bring the exact error message to class or office hours.

#### Terminal introduction

The terminal is part of the required workflow in this course. Even when you write code in an IDE, you should be able to navigate to a folder, inspect its contents, compile a Java file, and run the resulting program from the terminal.

- [ ] Watch the installation and walkthrough video used by COMP141 students: **[link](https://luc.hosted.panopto.com/Panopto/Pages/Viewer.aspx?id=29fb3f6a-024c-4461-8377-b3410009f54e&start=0)**
- [ ] Practice the commands appropriate to your system: `pwd`, `ls` (or `dir`), `cd`, and `mkdir`...
- [ ] Make sure you know how to open a terminal directly in a project folder.

#### What compiling from the terminal looks like

If a folder contains `Hello272.java`, the basic workflow is:

```bash
javac Hello272.java
java Hello272
```

- `javac` compiles the source file and normally creates `Hello272.class`.
- `java` starts the Java Virtual Machine and runs the compiled class.

We will practice this together. For now, your goal is simply to confirm that `java` and `javac` are available.
